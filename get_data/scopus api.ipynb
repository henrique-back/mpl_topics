# 1. Get data from SCOPUS API

```python
import pandas as pd

data = pd.read_csv('../get_data/scopus_data.csv')
print("Length of dataset retrieved from SCOPUS", len(data))

# Remove duplicates based on DOI. Papers without a DOI are all kept:
# drop_duplicates treats missing DOIs as equal and would keep only one of them.
dup_doi = data['DOI'].notna() & data.duplicated(subset=['DOI'])
print("Number of duplicates based on DOI:", dup_doi.sum())
print("Papers without a DOI (kept):", data['DOI'].isna().sum())
data = data[~dup_doi]

#Remove instances with no abstract available
data.dropna(subset=['Abstract'], inplace=True)
data = data[~data['Abstract'].str.contains('No abstract available')]
data.reset_index(drop=True, inplace=True)

# Just to make sure, filter DataFrame based on empty strings in the "Abstract" column
filtered_data = data[data['Abstract'].str.strip() == '']
indices = filtered_data.index
data.drop(indices, inplace=True)

# Remove "Corrigendum" instances
data = data[~data['Title'].str.contains('Corrigendum')]
data.reset_index(drop=True, inplace=True)

#Save
data.to_csv('../get_data/alldata.csv', index=False)
print("Final data set length is:", len(data))
print(data.shape)
```
